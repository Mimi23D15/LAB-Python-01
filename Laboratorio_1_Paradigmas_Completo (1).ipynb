### 1.1 Cuenta bancaria
El estado inicial son los valores de `saldo`, `deposito` y `retiro` guardados en memoria. Cada asignación actualiza una variable: al depositar, `saldo` aumenta; al retirar, disminuye y se muestra el **saldo antes del interés**. Luego `saldo_final` guarda el saldo con el interés del 2 % aplicado y se muestra el **saldo final con interés**. Con los valores del ejemplo, el saldo antes del interés es $570000 y el saldo final es $581400.

```python
saldo = float(input("Saldo inicial (ejemplo 500000): "))
deposito = float(input("Monto a depositar (ejemplo 150000): "))
retiro = float(input("Monto a retirar (ejemplo 80000): "))
interes = 0.02

saldo = saldo + deposito
saldo = saldo - retiro
print(f"Saldo antes del interés: ${saldo:.2f}")

saldo_final = saldo + saldo * interes
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

**Explicación: ¿qué representan el estado inicial y final y cómo cambia la memoria?**

El **estado inicial** es el conjunto de valores que hay guardados en las variables antes de operar: `saldo`, `deposito`, `retiro` e `interes`. `input()` recibe texto, y `float()` lo transforma en números para hacer operaciones.

- `saldo = saldo + deposito` cambia el valor almacenado en `saldo`, sumándole el depósito.
- `saldo = saldo - retiro` vuelve a actualizar `saldo`, restándole el retiro. Este es el saldo **antes de aplicar el interés**.
- `print(...)` muestra ese saldo, pero no cambia sus valores.
- `saldo_final = saldo + saldo * interes` crea una variable con el saldo después de agregarle el 2 %. `saldo` sigue guardando la cantidad anterior al interés.

El **estado final** incluye el saldo antes de interés en `saldo` y el saldo con interés en `saldo_final`. Por ejemplo, con 500000, 150000 y 80000, obtenemos **570000 antes del interés** y **581400 después**.

### 1.3 Inventario
Las operaciones se realizan de forma secuencial. El valor total se calcula **antes** de subir los precios, según el orden pedido; también se muestra después para comparar.

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10}
]
print("Primer producto:", inventario[0]["nombre"])
inventario.append({"nombre": "Arroz", "precio": 4200, "stock": 8})
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
valor_total = 0
for producto in inventario:
    valor_total += producto["precio"] * producto["stock"]
print("Valor total antes del aumento:", valor_total)
print("Productos con stock menor a 10:")
for producto in inventario:
    if producto["stock"] < 10:
        print(producto)
for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)
buscado = input("Nombre del producto a buscar: ").strip().lower()
encontrado = False
for producto in inventario:
    if producto["nombre"].lower() == buscado:
        print("Producto encontrado:", producto)
        encontrado = True
        break
if not encontrado:
    print("Ese producto no está en el inventario")
indice = 0
while indice < len(inventario):
    if inventario[indice]["stock"] == 0:
        inventario.pop(indice)
    else:
        indice += 1
# Ordenamiento ascendente por precio con ciclos (sin lambda ni funciones clave).
for i in range(len(inventario)):
    for j in range(i + 1, len(inventario)):
        if inventario[i]["precio"] > inventario[j]["precio"]:
            auxiliar = inventario[i]
            inventario[i] = inventario[j]
            inventario[j] = auxiliar
print("Inventario ordenado por precio:")
for producto in inventario:
    print(producto)
```

### 1.4 Tuplas y recorrido
Las tuplas son inmutables: una asignación como `rectangulo[0] = 10` produce `TypeError: 'tuple' object does not support item assignment`. Se captura el error para que el cuaderno continúe. La distancia entre dos puntos usa Pitágoras.

```python
import math
punto = (3, 7)
print("Punto:", punto)
rectangulo = (5, 4)
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
print("Área:", area, "Perímetro:", perimetro)
try:
    rectangulo[0] = 10
except TypeError as error:
    print("Error al modificar la tupla:", error)
posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0
for i in range(1, len(posiciones)):
    dx = posiciones[i][0] - posiciones[i-1][0]
    dy = posiciones[i][1] - posiciones[i-1][1]
    distancia_total += math.sqrt(dx * dx + dy * dy)
print("Distancia total recorrida:", distancia_total)
```

**Explicación: ¿qué error aparece al intentar cambiar el ancho de la tupla y por qué?**

Aparece `TypeError: 'tuple' object does not support item assignment`. Esto ocurre porque las tuplas son **inmutables**: después de crear `rectangulo = (5, 4)`, no se puede reemplazar únicamente uno de sus elementos mediante `rectangulo[0] = 10`. Si se necesitara otro ancho, habría que crear una nueva tupla y asignarla a `rectangulo`.

```python
suma_pares = 0
suma_impares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i
    print("Número:", i, "Estado:", suma_pares, suma_impares)
estado_final = {"suma_pares": suma_pares, "suma_impares": suma_impares}
print("Estado final:", estado_final)
```

Solo cambia un acumulador por vuelta. Al finalizar, `estado_final` guarda los resultados: `{'suma_pares': 30, 'suma_impares': 25}`.


### 2.5 Inventario con continue y max
Se omiten entradas negativas. `max()` identifica el precio más alto y, si hay empate, se conserva el primero encontrado.

### 4.1 Alcance local y global
El código original causa `UnboundLocalError` porque Python interpreta `contador_global` como variable local dentro de `incrementar`, pero se intenta leer antes de asignarla. La versión que recibe y devuelve valores es más fácil de probar, porque no depende de un estado global compartido.

```python
contador_global = 0
def incrementar_incorrecto():
    contador_global = contador_global + 1
try:
    incrementar_incorrecto()
except UnboundLocalError as error:
    print("Error original:", error)

contador_global = 0
def incrementar_global():
    global contador_global
    contador_global = contador_global + 1
incrementar_global()
print("Usando global:", contador_global)

def incrementar_por_parametro(valor):
    valor = valor + 1
    return valor
contador = 0
contador = incrementar_por_parametro(contador)
print("Usando parámetro y retorno:", contador)
```

**Explicación del error y de las dos soluciones**

El código original produce `UnboundLocalError`, porque Python trata `contador_global` como una variable **local** al ver una asignación dentro de la función, e intenta leer esa variable local antes de que tenga un valor.

Con `global contador_global`, la función modifica directamente la variable definida fuera de ella. En la otra versión, la función recibe un número, le suma 1 y devuelve el resultado con `return`; después el programa actualiza la variable con ese valor.

**¿Cuál es más fácil de probar?** La versión que recibe el valor y lo retorna, porque podemos darle entradas concretas y comprobar su salida sin depender de una variable global que otras partes del programa podrían modificar.


### 4.2 Mutabilidad, reasignación e id()
Predicción: se imprime `[3.5, 4.0, 4.8]`, porque `append` modifica la lista compartida y `notas = []` solo reasigna el nombre local. En la versión corregida, `clear()` modifica el mismo objeto.

```python
def agregar_nota(notas, nueva):
    notas.append(nueva)
def reiniciar_incorrecto(notas):
    print("id antes de reasignar dentro de la función:", id(notas))
    notas = []
    print("id después de reasignar dentro de la función:", id(notas))
def reiniciar(notas):
    print("id dentro de reiniciar antes de clear:", id(notas))
    notas.clear()
    print("id dentro de reiniciar después de clear:", id(notas))
mis_notas = [3.5, 4.0]
print("id original:", id(mis_notas))
agregar_nota(mis_notas, 4.8)
reiniciar_incorrecto(mis_notas)
print("Resultado del código original:", mis_notas)
print("id después de función incorrecta:", id(mis_notas))
reiniciar(mis_notas)
print("Lista vaciada:", mis_notas)
print("id final:", id(mis_notas))
```

**Explicación: predicción, mutabilidad, reasignación e `id()`**

Antes de corregir `reiniciar`, el programa original imprime **`[3.5, 4.0, 4.8]`**. `agregar_nota` usa `append()`, que modifica el contenido de la **misma lista**. En cambio, `notas = []` crea una nueva lista y hace que la variable local `notas` apunte a ella; `mis_notas`, fuera de la función, sigue apuntando a la lista original.

`id()` permite comprobarlo: antes de reasignar, el identificador local coincide con el de `mis_notas`; después de `notas = []`, es distinto. Con la versión corregida `notas.clear()`, se vacía la lista original y su `id()` se mantiene, porque se modificó el mismo objeto y no se creó otro.


### 4.3 Cajero automático
Los movimientos incluyen tipo, monto y saldo resultante. Consultar se registra como operación exitosa de monto 0; los depósitos y retiros inválidos no cambian el estado.

```python
cuenta = {"titular": "Ana", "saldo": 200000, "movimientos": []}
def registrar(cuenta, tipo, monto):
    cuenta["movimientos"].append({"tipo": tipo, "monto": monto, "saldo": cuenta["saldo"]})
def consultar(cuenta):
    print("Saldo disponible:", cuenta["saldo"])
    registrar(cuenta, "Consulta", 0)
def depositar(cuenta, monto):
    if monto <= 0:
        print("El depósito debe ser positivo")
    else:
        cuenta["saldo"] += monto
        registrar(cuenta, "Depósito", monto)
        print("Depósito realizado")
def retirar(cuenta, monto):
    if monto <= 0:
        print("El retiro debe ser positivo")
    elif monto > cuenta["saldo"]:
        print("Fondos insuficientes; no se modificó el saldo")
    else:
        cuenta["saldo"] -= monto
        registrar(cuenta, "Retiro", monto)
        print("Retiro realizado")
while True:
    opcion = input("Operación (consultar/depositar/retirar/salir): ").strip().lower()
    if opcion == "salir":
        break
    elif opcion == "consultar":
        consultar(cuenta)
    elif opcion == "depositar" or opcion == "retirar":
        try:
            monto = float(input("Monto: "))
            if opcion == "depositar":
                depositar(cuenta, monto)
            else:
                retirar(cuenta, monto)
        except ValueError:
            print("Debe ingresar un monto numérico")
    else:
        print("Opción no válida")
print("\nEXTRACTO DE MOVIMIENTOS")
print(f"{'Tipo':<14} {'Monto':>12} {'Saldo':>14}")
print("-" * 42)
for movimiento in cuenta["movimientos"]:
    print(f"{movimiento['tipo']:<14} {movimiento['monto']:>12.2f} {movimiento['saldo']:>14.2f}")
print("Saldo final:", cuenta["saldo"])
```

### 5.1 Adivina el número
Con búsqueda binaria, cada intento descarta aproximadamente la mitad de las posibilidades. Siete preguntas bastan para distinguir hasta `2^7 = 128` opciones, más que los 100 números del juego. El enunciado dice «búsqueda binaria del Ejercicio 5», pero no incluye otro apartado independiente con ese nombre.

```python
import random
ganadas = 0
perdidas = 0
jugar = "s"
while jugar == "s":
    secreto = random.randint(1, 100)
    historial = []
    adivino = False
    for intento in range(1, 8):
        try:
            numero = int(input(f"Intento {intento}/7. Número entre 1 y 100: "))
        except ValueError:
            print("Entrada inválida. Este intento se cuenta.")
            historial.append("inválido")
            continue
        historial.append(numero)
        if numero == secreto:
            print("¡Acertaste!")
            adivino = True
            break
        elif numero < secreto:
            print("El número secreto es mayor")
        else:
            print("El número secreto es menor")
    if adivino:
        ganadas += 1
    else:
        perdidas += 1
        print("El número era:", secreto)
    print("Historial:", historial)
    print("Ganadas:", ganadas, "Perdidas:", perdidas)
    jugar = input("¿Jugar otra vez? (s/n): ").strip().lower()
```

**Explicación: ¿por qué son suficientes 7 intentos con una estrategia óptima?**

Si probamos siempre el número de la **mitad del intervalo** posible, cada respuesta de «mayor» o «menor» permite descartar aproximadamente la mitad de los candidatos. Después de 7 decisiones podemos distinguir hasta `2 ** 7 = 128` posibilidades; como solo hay 100 números, **7 intentos son suficientes** cuando se eligen correctamente los números. Esta estrategia se conoce como **búsqueda binaria**. Adivinar números al azar no garantiza acertar en 7 intentos.


### 5.2 Juego de la vida de Conway
El tablero nuevo evita que los cambios de las celdas procesadas alteren el conteo de vecinos de las siguientes celdas durante la misma generación. La configuración inicial es un patrón oscilador de tres celdas.

```python
filas = 10
columnas = 10
tablero = []
for fila in range(filas):
    linea = []
    for columna in range(columnas):
        linea.append(0)
    tablero.append(linea)
tablero[4][3] = 1
tablero[4][4] = 1
tablero[4][5] = 1

def contar_vecinos(tablero, fila, columna):
    vecinos = 0
    for cambio_fila in range(-1, 2):
        for cambio_columna in range(-1, 2):
            if cambio_fila == 0 and cambio_columna == 0:
                continue
            otra_fila = fila + cambio_fila
            otra_columna = columna + cambio_columna
            if 0 <= otra_fila < len(tablero) and 0 <= otra_columna < len(tablero[0]):
                vecinos += tablero[otra_fila][otra_columna]
    return vecinos

def siguiente_generacion(tablero):
    nuevo = []
    for fila in range(len(tablero)):
        linea = []
        for columna in range(len(tablero[0])):
            vecinos = contar_vecinos(tablero, fila, columna)
            viva = tablero[fila][columna]
            if viva == 1 and (vecinos == 2 or vecinos == 3):
                linea.append(1)
            elif viva == 0 and vecinos == 3:
                linea.append(1)
            else:
                linea.append(0)
        nuevo.append(linea)
    return nuevo

def mostrar_tablero(tablero):
    for fila in tablero:
        texto = ""
        for celda in fila:
            if celda == 1:
                texto += "■ "
            else:
                texto += "· "
        print(texto)

for generacion in range(1, 6):
    print("Generación", generacion)
    mostrar_tablero(tablero)
    tablero = siguiente_generacion(tablero)
    print()
```

**Explicación: ¿por qué calcular la siguiente generación en un tablero nuevo?**

Todas las celdas de una generación deben decidir si sobreviven, mueren o nacen **mirando el mismo tablero original**. Si modificáramos ese tablero mientras lo recorremos, las siguientes celdas contarían vecinos que ya cambiaron y los resultados serían incorrectos. Por eso creamos `nuevo_tablero`, calculamos allí todos los estados y, solamente al terminar, reemplazamos el tablero anterior.


# Laboratorio 1 — Programación imperativa
## Ejercicio 6
Código diseñado para ejecutarse en Google Colab. Ejecuta cada celda en orden.

### 6.1 Validación de entradas
`except` se ejecuta al haber `ValueError`, `else` cuando la conversión funciona, y `finally` siempre, incluso después de un error. Las entradas negativas se saltan con `continue`.

```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ").strip()
    if nombre.lower() == "fin":
        break
    while True:
        try:
            precio = float(input("Precio: "))
        except ValueError:
            print("Precio inválido, inténtelo nuevamente")
        else:
            print("Precio numérico recibido")
            break
        finally:
            print("Finalizó un intento de lectura del precio")
    while True:
        try:
            cantidad = int(input("Cantidad: "))
        except ValueError:
            print("Cantidad inválida, inténtelo nuevamente")
        else:
            break
    if precio < 0 or cantidad < 0:
        print("Producto omitido por precio o cantidad negativa")
        continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})
total = 0
for producto in productos:
    total += producto["precio"] * producto["cantidad"]
print("Inventario:", productos)
print("Valor total:", total)
```

**Explicación: ¿cuándo se ejecutan `except`, `else` y `finally`?**

- `try` contiene la conversión que podría fallar, como `float(input(...))`.
- `except ValueError` se ejecuta cuando el usuario escribe algo que no puede convertirse en número; se informa del error y el `while` vuelve a pedir el dato.
- `else` se ejecuta cuando el bloque `try` termina sin error; en este caso confirma la entrada válida y permite salir del ciclo de lectura.
- `finally` se ejecuta **siempre**, ocurra o no un error. Aquí indica que finalizó ese intento de lectura.

De esta forma el inventario solo incorpora productos con precio y cantidad numéricos válidos y no negativos.


### 6.2 Persistencia CSV y JSON
Esta celda tiene su propio inventario de ejemplo para que se ejecute por separado. Los archivos se crean en el directorio actual de Colab (normalmente `/content`), que es temporal. Para conservarlos hay que descargarlos o guardarlos en Drive. CSV guarda texto por filas y necesita conversiones de tipos al leerlo; JSON conserva listas, diccionarios y tipos numéricos directamente. Esta versión básica de CSV asume nombres de productos sin comas ni saltos de línea.

```python
import json
inventario = [
    {"nombre": "Pan", "precio": 1500.0, "cantidad": 3},
    {"nombre": "Leche", "precio": 3500.0, "cantidad": 2}
]

def guardar_csv(inventario, archivo):
    with open(archivo, "w", encoding="utf-8") as salida:
        salida.write("nombre,precio,cantidad\n")
        for producto in inventario:
            salida.write(f"{producto['nombre']},{producto['precio']},{producto['cantidad']}\n")

def leer_csv(archivo):
    productos = []
    try:
        with open(archivo, "r", encoding="utf-8") as entrada:
            primera = True
            for linea in entrada:
                if primera:
                    primera = False
                    continue
                datos = linea.strip().split(",")
                if len(datos) == 3:
                    productos.append({"nombre": datos[0], "precio": float(datos[1]), "cantidad": int(datos[2])})
    except FileNotFoundError:
        print("CSV no existe; inventario vacío")
    return productos

def guardar_json(inventario, archivo):
    with open(archivo, "w", encoding="utf-8") as salida:
        json.dump(inventario, salida, ensure_ascii=False, indent=2)

def leer_json(archivo):
    try:
        with open(archivo, "r", encoding="utf-8") as entrada:
            productos = json.load(entrada)
    except FileNotFoundError:
        print("JSON no existe; inventario vacío")
        productos = []
    return productos

print("CSV inexistente:", leer_csv("csv_inexistente.csv"))
print("JSON inexistente:", leer_json("json_inexistente.json"))
guardar_csv(inventario, "inventario.csv")
print("Desde CSV:", leer_csv("inventario.csv"))
guardar_json(inventario, "inventario.json")
print("Desde JSON:", leer_json("inventario.json"))
```

**Explicación: ¿qué ventajas tiene JSON frente a CSV?**

Con **CSV** escribimos cada producto como una línea de texto separada por comas. Al leerlo, debemos separar cada campo y convertir manualmente los precios y cantidades a `float` o `int`. Con **JSON**, `json.dump()` guarda directamente nuestra lista de diccionarios y `json.load()` recupera su estructura y sus números sin tener que reconstruir cada diccionario a mano. JSON también resulta más cómodo para datos anidados. CSV puede ser práctico para tablas sencillas que se van a abrir en programas de hojas de cálculo.
